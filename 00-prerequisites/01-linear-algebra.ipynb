## 2. 벡터 연산

### 2.1 기본 연산

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# 덧셈/뺄셈 (element-wise)
print(f"a + b = {a + b}")
print(f"a - b = {a - b}")

# 스칼라 곱
print(f"3 * a = {3 * a}")

# element-wise 곱 (Hadamard product)
print(f"a * b = {a * b}  (element-wise)")
```

### 2.2 내적 (Dot Product)

내적은 ML에서 가장 많이 쓰이는 연산 중 하나:
- 신경망의 뉴런 계산: `output = w · x + b`
- 유사도 측정: 코사인 유사도의 분자

$$\mathbf{a} \cdot \mathbf{b} = \sum_{i} a_i b_i$$

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# 내적 - 3가지 방법
dot1 = np.dot(a, b)
dot2 = a @ b
dot3 = np.sum(a * b)  # 직접 계산: element-wise 곱 후 합

print(f"np.dot(a, b) = {dot1}")
print(f"a @ b        = {dot2}")
print(f"sum(a * b)   = {dot3}")
print(f"직접 계산: 1*4 + 2*5 + 3*6 = {1*4 + 2*5 + 3*6}")
```

### 2.3 노름 (Norm)

벡터의 크기(길이)를 측정. ML에서 정규화(Regularization)에 사용.

- **L1 노름**: $\|\mathbf{x}\|_1 = \sum |x_i|$ → Lasso 정규화
- **L2 노름**: $\|\mathbf{x}\|_2 = \sqrt{\sum x_i^2}$ → Ridge 정규화, 유클리드 거리

```python
v = np.array([3, 4])

# L1 노름 (Manhattan distance)
l1 = np.linalg.norm(v, ord=1)
l1_manual = np.sum(np.abs(v))
print(f"L1 노름: {l1} (직접: {l1_manual})")

# L2 노름 (Euclidean distance)
l2 = np.linalg.norm(v, ord=2)
l2_manual = np.sqrt(np.sum(v ** 2))
print(f"L2 노름: {l2} (직접: {l2_manual})")
print(f"→ [3, 4] 벡터의 길이는 {l2} (3-4-5 직각삼각형)")
```

### 2.4 코사인 유사도 (Cosine Similarity)

두 벡터의 방향이 얼마나 비슷한지 측정. NLP에서 단어/문장 유사도에 핵심적으로 사용.

$$\cos(\theta) = \frac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{a}\| \|\mathbf{b}\|}$$

- 1에 가까우면: 같은 방향 (유사)
- 0이면: 직교 (무관)
- -1이면: 반대 방향

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# 같은 방향
v1 = np.array([1, 0])
v2 = np.array([3, 0])
print(f"같은 방향: {cosine_similarity(v1, v2):.2f}")

# 직교
v3 = np.array([0, 1])
print(f"직교:      {cosine_similarity(v1, v3):.2f}")

# 반대 방향
v4 = np.array([-1, 0])
print(f"반대 방향: {cosine_similarity(v1, v4):.2f}")

# 비슷한 방향
v5 = np.array([1, 0.1])
print(f"비슷한 방향: {cosine_similarity(v1, v5):.4f}")
```

### 2.5 벡터 시각화

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 왼쪽: 벡터 덧셈 시각화
ax = axes[0]
a = np.array([2, 1])
b = np.array([1, 2])
ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='red', label='b')
ax.quiver(0, 0, (a+b)[0], (a+b)[1], angles='xy', scale_units='xy', scale=1, color='green', label='a+b')
ax.set_xlim(-0.5, 4.5)
ax.set_ylim(-0.5, 4.5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title('Vector Addition')

# 오른쪽: 코사인 유사도 시각화
ax = axes[1]
vectors = {
    'a (기준)': np.array([1, 0]),
    'b (유사)': np.array([0.9, 0.2]),
    'c (직교)': np.array([0, 1]),
    'd (반대)': np.array([-1, 0.1]),
}
colors = ['blue', 'green', 'orange', 'red']
for (name, v), color in zip(vectors.items(), colors):
    v_norm = v / np.linalg.norm(v)  # 단위벡터로 정규화
    sim = cosine_similarity(vectors['a (기준)'], v)
    ax.quiver(0, 0, v_norm[0], v_norm[1], angles='xy', scale_units='xy', scale=1,
              color=color, label=f'{name}: cos={sim:.2f}')
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-0.5, 1.5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=9)
ax.set_title('Cosine Similarity')

plt.tight_layout()
plt.show()
```

## 3. 행렬 연산

### 3.1 행렬 곱셈

신경망의 핵심 연산. 입력 데이터 X에 가중치 W를 곱하는 것이 행렬 곱.

$$(AB)_{ij} = \sum_k A_{ik} B_{kj}$$

조건: A의 열 수 == B의 행 수

```python
# 행렬 곱셈
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# 행렬 곱 (dot product)
C = A @ B  # 또는 np.matmul(A, B)
print(f"A @ B =\n{C}")

# 직접 계산으로 확인
print(f"\n직접 계산:")
print(f"C[0,0] = 1*5 + 2*7 = {1*5 + 2*7}")
print(f"C[0,1] = 1*6 + 2*8 = {1*6 + 2*8}")
print(f"C[1,0] = 3*5 + 4*7 = {3*5 + 4*7}")
print(f"C[1,1] = 3*6 + 4*8 = {3*6 + 4*8}")

# 주의: element-wise 곱과 다름!
print(f"\nA * B (element-wise) =\n{A * B}")
```

### 3.2 신경망에서의 행렬 곱

뉴런 하나의 계산: `output = w · x + b`

이것을 여러 뉴런, 여러 데이터에 대해 한꺼번에 계산하면 → 행렬 곱

```python
# 예시: 3개 특성 → 2개 뉴런
# 입력: 4개 데이터, 각각 3개 특성
X = np.random.randn(4, 3)   # (batch_size=4, features=3)
W = np.random.randn(3, 2)   # (features=3, neurons=2)
b = np.random.randn(2)      # (neurons=2)

# 신경망 forward pass (한 레이어)
output = X @ W + b           # (4, 3) @ (3, 2) + (2,) = (4, 2)

print(f"입력 X shape:   {X.shape}  (4개 데이터, 3개 특성)")
print(f"가중치 W shape: {W.shape}  (3개 입력, 2개 뉴런)")
print(f"출력 shape:     {output.shape}  (4개 데이터, 2개 출력)")
print(f"\n출력:\n{output}")
```

### 3.3 전치, 역행렬, 행렬식

```python
A = np.array([[1, 2],
              [3, 4]])

# 전치 (Transpose): 행과 열을 뒤바꿈
print(f"A =\n{A}")
print(f"A^T =\n{A.T}")

# 행렬식 (Determinant): 행렬이 공간을 얼마나 확대/축소하는지
det = np.linalg.det(A)
print(f"\ndet(A) = {det:.1f}")
print(f"직접 계산: 1*4 - 2*3 = {1*4 - 2*3}")

# 역행렬 (Inverse): A @ A_inv = I (단위행렬)
A_inv = np.linalg.inv(A)
print(f"\nA^(-1) =\n{A_inv}")
print(f"\nA @ A^(-1) =\n{np.round(A @ A_inv, 10)}  (= 단위행렬)")
```

### 3.4 Broadcasting

NumPy/PyTorch에서 shape이 다른 배열 간 연산을 자동으로 맞춰주는 기능.
ML에서 bias 더하기 등에 항상 사용됨.

```python
# (4, 3) 행렬에 (3,) 벡터를 더하기 → 각 행에 벡터가 더해짐
X = np.ones((4, 3))
b = np.array([10, 20, 30])

result = X + b  # b가 4번 반복되어 더해짐 (broadcasting)
print(f"X (4x3):\n{X}")
print(f"b (3,): {b}")
print(f"X + b (4x3):\n{result}")
```

## 4. 고유값과 고유벡터 (Eigenvalue & Eigenvector)

행렬 A를 곱했을 때 **방향이 변하지 않고** 크기만 변하는 벡터가 고유벡터.

$$A\mathbf{v} = \lambda\mathbf{v}$$

- $\mathbf{v}$: 고유벡터 (방향)
- $\lambda$: 고유값 (크기 변화 배수)

**ML에서의 활용**: PCA에서 데이터의 주요 방향(분산이 큰 방향)을 찾을 때 사용

```python
A = np.array([[2, 1],
              [1, 2]])

eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"고유값: {eigenvalues}")
print(f"고유벡터:\n{eigenvectors}")
print(f"(각 열이 하나의 고유벡터)")

# 검증: A @ v = lambda * v
for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    lhs = A @ v
    rhs = lam * v
    print(f"\n고유벡터 {i+1}: v = {v}")
    print(f"  A @ v     = {lhs}")
    print(f"  lambda*v  = {rhs}")
    print(f"  같은가? {np.allclose(lhs, rhs)}")
```

### 고유벡터 시각화

행렬 A가 벡터에 곱해질 때, 대부분의 벡터는 방향이 바뀐다.
하지만 고유벡터는 방향이 유지되고 크기(고유값)만 변한다.

```python
A = np.array([[2, 1],
              [1, 2]])
eigenvalues, eigenvectors = np.linalg.eig(A)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 왼쪽: 일반 벡터 → 방향이 바뀜
ax = axes[0]
v = np.array([1, 0.3])
Av = A @ v
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='blue', label=f'v = {v}')
ax.quiver(0, 0, Av[0], Av[1], angles='xy', scale_units='xy', scale=1, color='red', label=f'Av = [{Av[0]:.1f}, {Av[1]:.1f}]')
ax.set_xlim(-0.5, 3.5)
ax.set_ylim(-0.5, 3.5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title('General vector: direction CHANGES')

# 오른쪽: 고유벡터 → 방향 유지, 크기만 변함
ax = axes[1]
v = eigenvectors[:, 0]
Av = A @ v
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='blue', label=f'v = [{v[0]:.2f}, {v[1]:.2f}]')
ax.quiver(0, 0, Av[0], Av[1], angles='xy', scale_units='xy', scale=1, color='red', label=f'Av = [{Av[0]:.2f}, {Av[1]:.2f}] (= {eigenvalues[0]:.0f}v)')
ax.set_xlim(-0.5, 3.5)
ax.set_ylim(-0.5, 3.5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title(f'Eigenvector: direction SAME, scaled by λ={eigenvalues[0]:.0f}')

plt.tight_layout()
plt.show()
```

## 5. PCA (Principal Component Analysis)

ML에서 선형대수가 실전으로 쓰이는 대표적 예시.

고차원 데이터를 **분산이 가장 큰 방향**(=고유벡터)으로 압축하여 차원을 줄인다.

1. 데이터를 평균 0으로 중심 이동
2. 공분산 행렬 계산
3. 고유값 분해 → 고유값이 큰 축이 주성분
4. 상위 k개 고유벡터로 차원 축소

```python
np.random.seed(42)

# 2D 데이터 생성 (상관관계 있는 데이터)
mean = [2, 3]
cov = [[2, 1.5],
       [1.5, 1.5]]
data = np.random.multivariate_normal(mean, cov, 200)

# Step 1: 평균 중심 이동
data_centered = data - data.mean(axis=0)

# Step 2: 공분산 행렬
cov_matrix = np.cov(data_centered.T)
print(f"공분산 행렬:\n{cov_matrix}")

# Step 3: 고유값 분해
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# 고유값 큰 순서로 정렬
idx = eigenvalues.argsort()[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print(f"\n고유값: {eigenvalues}")
print(f"분산 설명 비율: {eigenvalues / eigenvalues.sum() * 100}%")
print(f"→ 첫 번째 주성분이 전체 분산의 {eigenvalues[0] / eigenvalues.sum() * 100:.1f}%를 설명")
```

```python
# PCA 시각화
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 왼쪽: 원본 데이터 + 주성분 방향
ax = axes[0]
ax.scatter(data_centered[:, 0], data_centered[:, 1], alpha=0.3, s=10)
for i, (val, vec) in enumerate(zip(eigenvalues, eigenvectors.T)):
    ax.quiver(0, 0, vec[0]*val, vec[1]*val, angles='xy', scale_units='xy', scale=1,
              color=['red', 'blue'][i], linewidth=2,
              label=f'PC{i+1} (λ={val:.2f})')
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title('Original Data + Principal Components')

# 오른쪽: 1차원으로 축소 (PC1에 투영)
ax = axes[1]
pc1 = eigenvectors[:, 0]
projected = data_centered @ pc1  # 1D로 축소
reconstructed = np.outer(projected, pc1)  # 다시 2D로 복원

ax.scatter(data_centered[:, 0], data_centered[:, 1], alpha=0.2, s=10, label='Original')
ax.scatter(reconstructed[:, 0], reconstructed[:, 1], alpha=0.3, s=10, color='red', label='Projected to PC1')
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title('2D → 1D (PCA Dimensionality Reduction)')

plt.tight_layout()
plt.show()
```

## 6. 선형 변환 시각화

행렬 곱은 기하학적으로 **공간을 변환**하는 것.
3Blue1Brown 영상의 핵심 내용을 코드로 확인.

```python
def plot_transformation(ax, A, title):
    """행렬 A에 의한 단위 격자 변환을 시각화"""
    # 단위 격자 점
    t = np.linspace(-2, 2, 20)
    grid = []
    for i in t:
        for j in t:
            grid.append([i, j])
    grid = np.array(grid)

    # 변환 적용
    transformed = (A @ grid.T).T

    ax.scatter(transformed[:, 0], transformed[:, 1], s=3, alpha=0.5)
    # 기저벡터 변환 결과
    ax.quiver(0, 0, A[0, 0], A[1, 0], angles='xy', scale_units='xy', scale=1, color='red', linewidth=2, label='e1 transformed')
    ax.quiver(0, 0, A[0, 1], A[1, 1], angles='xy', scale_units='xy', scale=1, color='blue', linewidth=2, label='e2 transformed')
    ax.set_xlim(-4, 4)
    ax.set_ylim(-4, 4)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.set_title(f"{title}\ndet={np.linalg.det(A):.1f}")
    ax.legend(fontsize=8)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))

transforms = {
    'Identity': np.eye(2),
    'Scale (2x)': np.array([[2, 0], [0, 2]]),
    'Rotation (45°)': np.array([[np.cos(np.pi/4), -np.sin(np.pi/4)],
                                [np.sin(np.pi/4), np.cos(np.pi/4)]]),
    'Shear': np.array([[1, 1], [0, 1]]),
}

for ax, (name, A) in zip(axes, transforms.items()):
    plot_transformation(ax, A, name)

plt.tight_layout()
plt.show()
```
